# Análisis de ventas y generación de insights
En este notebook se realiza el análisis descriptivo y estadístico del dataset final, con el objetivo de obtener insights de negocio y preparar la información para su visualización en Power BI.

# Importaciones

```python
import pandas as pd
import numpy as np
```

# Carga del dataset final

```python
df = pd.read_csv(
    "../data/processed_data/retail_final_dataset.csv",
    parse_dates=["Order Date", "Ship Date"]
)
```

# KPIs principales 

```python
total_sales = df["Sales"].sum()
avg_order_value = df["Sales"].mean()
total_orders = df["Order ID"].nunique()
total_customers = df["Customer ID"].nunique()
total_quantity = df["Quantity"].sum()

total_sales, avg_order_value, total_orders, total_customers, total_quantity
```

# Análisis temporal de ventas
Variables temporales.
Ventas por año.

```python
df["Order Year"] = df["Order Date"].dt.year
df["Order Month"] = df["Order Date"].dt.month
df["Order Month Name"] = df["Order Date"].dt.month_name()

sales_by_year = df.groupby("Order Year")["Sales"].sum().reset_index()
sales_by_year
```

# Análisis por categoría y producto

## Venta por categoría

```python
sales_by_category = (
    df.groupby("Category")["Sales"]
    .sum()
    .sort_values(ascending=False)
    .reset_index()
)
```

## Top 10 

```python
top_products = (
    df.groupby("Product Name")["Sales"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
    .reset_index()
)

top_products
```

## Análisis geográfico
Ventas por pais

```python
sales_by_country = (
    df.groupby("Country")["Sales"]
    .sum()
    .sort_values(ascending=False)
    .reset_index()
)

sales_by_country.head(10)
```

## Segmentación de clientes

```python
sales_by_segment = (
    df.groupby("Segment")["Sales"]
    .sum()
    .reset_index()
)

sales_by_segment
```

## Preparación de datasets para Power BI

```python
sales_by_year.to_csv(
    "../data/processed_data/sales_by_year.csv",
    index=False
)

sales_by_category.to_csv(
    "../data/processed_data/sales_by_category.csv",
    index=False
)

sales_by_country.to_csv(
    "../data/processed_data/sales_by_country.csv",
    index=False
)

sales_by_segment.to_csv(
    "../data/processed_data/sales_by_segment.csv",
    index=False
)
```

## Conclusiones del análisis
El análisis muestra patrones claros en la evolucion temporal de las ventas, una concentración de ingresos en determinadas categorías y países, y diferencias significativas entre segmentos de clientes. Estos insights permiten orientar estrategias comeciales y de optimizació de catálogo.
