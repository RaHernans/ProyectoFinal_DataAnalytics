# Limpieza y transformación de datos
Se realizan las tareas de limpieza, transformación y unión de los conjuntos de datos de ventas y países, con el objetivo de generar un dataset final listo para el análisis y la visualización. 

# Importaciones

```python
import pandas as pd
import numpy as np
```

# Carga de datos

```python
sales = pd.read_csv(
    "../data/raw_data/global_superstore.csv",
    encoding="latin1"
)

countries = pd.read_csv("../data/raw_data/countries.csv")
```

# Conversión de fechas 
Las columnas de fecha se convierten a formato datetime para permitir análisis temporales y agregaciones posteriores.

```python
sales["Order Date"] = pd.to_datetime(
    sales["Order Date"],
    dayfirst=True
)

sales["Ship Date"] = pd.to_datetime(
    sales["Ship Date"],
    dayfirst=True
)
```

```python
sales[["Order Date", "Ship Date"]].dtypes
```

# Tratamiento de valores nulos
Se analizan y tratan los valores nulos detectados durante el EDA inicial

## Código postal
Postal Code tiene muchos nulos y no es clave para el análisis global.

```python
sales["Postal Code"].isna().sum()
sales = sales.drop(columns=["Postal Code"])
```

# Normalización de variables categóricas
Se estandarizan los nombre de países para facilitar la unión entre datasets.

```python
sales["Country"] = sales["Country"].str.strip()
countries["Country"] = countries["Country"].str.strip()
```

# Integración de los datasets
Se realiza la unión entre el dataset de ventas y el dataset de países utilizando la columna Country como clave común.

```python
final_df = sales.merge(
    countries,
    how="left",
    on="Country"
)
```

## Validación del dataset final

```python
final_df.shape
final_df.info()
```

# Exportación del dataset final
El dataset transformado se guarda para su uso en análisis posteriores y para la construcción del dashoard en Power BI.

```python
final_df.to_csv(
    "../data/processed_data/retail_final_dataset.csv",
    index=False
)
```
